# 2.访问模型

```python
print("正在调用大模型...")

response = client.chat.completions.create(
    model="deepseek-chat",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "明天北京天气怎么样"}
    ],
    stream=False
)

print("成功调用大模型...")
```
